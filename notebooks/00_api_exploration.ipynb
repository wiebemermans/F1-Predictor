```python
import json

import requests

BASE = "https://api.jolpi.ca/ergast/f1"

# 1. Eén race: hoe ziet de structuur eruit?
data = requests.get(f"{BASE}/2023/1/results/", timeout=30).json()["MRData"]
print({k: data[k] for k in ("limit", "offset", "total")})
race = data["RaceTable"]["Races"][0]
print(race.keys())
print(json.dumps(race["Results"][0], indent=2))
```

```python
# 2. Paginering: telt limit rijen of races?
page = requests.get(f"{BASE}/2023/results/", params={"limit": 100}, timeout=30).json()["MRData"]
races = page["RaceTable"]["Races"]
print("total:", page["total"], "| races op pagina:", len(races))
print("rijen op pagina:", sum(len(r["Results"]) for r in races))
```

```python
# 3. Kwalificatie en sprint
q = requests.get(f"{BASE}/2023/1/qualifying/", timeout=30).json()["MRData"]
print(json.dumps(q["RaceTable"]["Races"][0]["QualifyingResults"][0], indent=2))

s = requests.get(f"{BASE}/2023/sprint/", timeout=30).json()["MRData"]
print(s["total"], s["RaceTable"]["Races"][0].keys())
```
